## 2. Target or proxy

I would use an observed search-performance outcome as the target. The target should represent whether a webpage shows weaker performance in a later observed period, rather than being a label created directly from the same features used for prediction. For this starter-data framing, I will use the observed engagement outcome as a proxy for page performance. The target should be measured after the input metrics so that the model learns from earlier signals to support a later decision.



```python
import pandas as pd

data_path = "/content/flyrank-ml-internship/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(data_path)

print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

## 3. Success metric

My primary success metric will be Precision@K. The model is useful if a high proportion of the top-ranked webpages are genuinely useful pages to prioritize for improvement. I would choose a practical K, such as the top 50 pages, because an SEO team has limited time and cannot act on every webpage at once. The metric should be evaluated against an observed later outcome.


```python
K = 50

print("Planned evaluation metric: Precision@K")
print("K =", K)
print("Interpretation: How many of the top", K, "ranked pages are relevant for improvement?")
```

## 5. Why ML beats a fixed rule here
A fixed rule such as "improve every page with CTR below a certain threshold" would be too simple because webpage performance depends on several signals, including impressions, average position, CTR, engagement rate, content characteristics, and search demand. These signals can interact, and the useful patterns may differ across webpages. ML can combine multiple observed signals to produce a priority ranking instead of relying on one manually chosen threshold. The output would be used as decision support to help SEO analysts prioritize which webpages to improve first, not as a guarantee of future search performance.
